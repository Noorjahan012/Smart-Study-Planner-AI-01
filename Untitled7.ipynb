```python
# 📚 SMART STUDY PLANNER WITH AI COACH

import pandas as pd
import math
import matplotlib.pyplot as plt

print("📚 SMART STUDY PLANNER 📚\n")

# ---------- INPUT ----------
n = int(input("Enter number of subjects: "))

subjects = []
hours = []
priority = []

for i in range(n):
    print(f"\nSubject {i+1}")
    sub = input("Enter subject name: ")
    h = int(input("Enter hours required: "))
    p = int(input("Enter priority (1-5, 5 = highest): "))

    subjects.append(sub)
    hours.append(h)
    priority.append(p)

# ---------- DATAFRAME ----------
df = pd.DataFrame({
    "Subject": subjects,
    "Hours": hours,
    "Priority": priority
})

# ---------- HEURISTIC SORT ----------
df = df.sort_values(by=["Priority", "Hours"], ascending=[False, False])

# ---------- DAILY LIMIT ----------
daily_limit = int(input("\nEnter how many hours you can study per day: "))

# ---------- ESTIMATION ----------
total_hours = sum(hours)
estimated_days = math.ceil(total_hours / daily_limit)

print("\n📊 ANALYSIS")
print(f"Total Study Hours: {total_hours}")
print(f"Estimated Days Required: {estimated_days} days\n")

# ---------- SCHEDULING ----------
schedule = []
day = 1
remaining_hours = df.copy()

print("📅 Generating Smart Schedule...\n")

while not remaining_hours.empty:
    day_hours = 0
    today_plan = []

    for index, row in remaining_hours.iterrows():
        if row["Hours"] <= (daily_limit - day_hours):
            today_plan.append((row["Subject"], row["Hours"]))
            day_hours += row["Hours"]
            remaining_hours = remaining_hours.drop(index)

    # Split if needed
    if not today_plan and not remaining_hours.empty:
        index = remaining_hours.index[0]
        row = remaining_hours.loc[index]
        today_plan.append((row["Subject"], daily_limit))
        remaining_hours.at[index, "Hours"] -= daily_limit

    schedule.append((day, today_plan))
    day += 1

# ---------- PRINT SCHEDULE ----------
for day, tasks in schedule:
    print(f"📌 Day {day}:")
    for sub, hrs in tasks:
        print(f"   - {sub} ({hrs} hrs)")
    print()

print(f"✅ Actual Days Taken by Scheduler: {len(schedule)} days")

# ---------- VISUALIZATION ----------
plt.figure()
plt.bar(df["Subject"], df["Hours"])
plt.title("Study Hours Distribution")
plt.xlabel("Subjects")
plt.ylabel("Hours")
plt.show()

# ---------- AI COACH ----------
print("\n🧠 AI STUDY COACH RECOMMENDATIONS\n")

def get_recommendations(subject, hrs, priority):
    s = subject.lower()

    time_slot = "Evening (after college/work)"
    method = "Revise + light practice"
    memory = "Use active recall (test yourself)"
    focus = "25 min study + 5 min break"
    food = "Water + nuts"

    if any(x in s for x in ["math", "physics", "accounts"]):
        time_slot = "Morning (fresh mind)"
        method = "Practice problems"
        memory = "Repeat formulas + solve questions"
        focus = "Deep focus (no phone)"
        food = "Banana + almonds"

    elif any(x in s for x in ["history", "biology", "theory"]):
        time_slot = "Afternoon"
        method = "Read + short notes"
        memory = "Use flashcards"
        focus = "Read and explain aloud"
        food = "Fruits + water"

    elif any(x in s for x in ["english", "language"]):
        time_slot = "Evening"
        method = "Read + write summaries"
        memory = "Use keywords & mind maps"
        focus = "Practice writing"
        food = "Dark chocolate (small)"

    if priority >= 4:
        focus += " | Start this first"

    if hrs >= 3:
        focus += " | Split into sessions"

    return time_slot, method, memory, focus, food


# ---------- PRINT AI SUGGESTIONS ----------
for day, tasks in schedule:
    print(f"\n📌 Day {day} Plan with AI Suggestions:")
    for sub, hrs in tasks:
        pr = int(df[df["Subject"] == sub]["Priority"].values[0])

        t, m, mem, f, food = get_recommendations(sub, hrs, pr)

        print(f"\n👉 {sub} ({hrs} hrs)")
        print(f"   🕒 Time: {t}")
        print(f"   📘 Method: {m}")
        print(f"   🧠 Memory: {mem}")
        print(f"   🎯 Focus: {f}")
        print(f"   🍎 Food: {food}")
```
